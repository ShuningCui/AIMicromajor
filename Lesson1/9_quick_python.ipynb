## <font color = #0288d1>字符串索引</font>
### 可以通过输入数字来访问第n个位置的字符。字符串中的每个字符都有一个位置编号，称为<font color="purple">**索引**</font>。第N个位置的索引紧跟在字符串后面的两个方括号[ ]之间
>### 注意在Python和大多数其他编程语言中，<font color="purple">**计数总是从0开始**</font>。所以位置1得到的是第2个字符，要获取字符串开头的字符，你需要访问位置0的字符。<font color="darkred">**如果访问的位置超出了界限 Python将引发IndexError。</font> 字符串中最大的索引总是比字符串的长度小1**。

```python
my_name = "CuiShuning"
my_name[9]
```

### 字符串也支持<font color="purple">**负索引**</font>，字符字符串中最后一个字符的索引是-1，对于“CuiShuning”来说是字母g，倒数第二个字符i的索引是-2，以此类推

```python
my_name[-2]
```

## <font color = #0288d1>字符串分割</font>
### 可以通过在方括号内的两个索引数之间插入冒号来提取字符串的一部分，称为子字符串
>### my_name[0:3]返回字符串的前3个字符，从索引0的字符开始，一直到索引3 **（<font color="darkred">注意不包括索引3的字符！</font>）**。字符串切片可能会令人困惑，因为切片返回的子字符串包含下标为第一个数字的字符，但不包含下标为第二个数字的字符。

```python
my_name = "CuiShuning"
my_name[0:3]
```

+ ### 如果开始的索引是0，则0可以省略
+ ### 如果省略了切片中的第二个下标，Python假定程序返回子字符串以指定的索引开始，以最后一个字符结束
+ ### 如果在切片中同时省略了第一个和第二个数字，就会得到一个以索引为0的字符开始，以最后一个字符结束的字符串。换句话说，在切片中省略这两个数字将返回整个字符串。
>### 需要注意的是，与字符串索引不同，试图在边界之前或之后进行切片时，Python<font color="darkred">**不会**</font>引发IndexError。任何不存在的索引都会被忽略，并<font color="purple">**返回整个字符串**</font>。试图得到一个整个范围都超出界限的切片时将<font color="purple">**返回空字符串""**</font>，而不是抛出一个错误。
>### 我们可以在切片中<font color="purple">**使用负数**</font>。负数切片的规则与正数切片的规则完全相同

```python
my_name[6:]
```

```python
my_name[:]
```

```python
my_name[0:20]
```

```python
my_name[-3:1]
```

```python
my_name[-4:]
```

## <font color = #0288d1>函数</font>
### 函数将代码分解为更小的块，非常适合程序重复使用的任务。每次程序需要执行任务时，不必编写相同的代码，只需调用函数即可。

### 执行函数的过程可以归纳为三个步骤:
### 1. <font color="purple">**函数调用**</font>，并将任何参数传递给函数。
### 2. <font color="purple">**函数执行**</font>，并对参数进行一些操作。 
### 3. <font color="purple">**函数返回**</font>，原始的函数调用被替换为返回值。

## <font color = #0288d1>编写自己的函数</font>
### 每个函数都有两个部分:
### 1. <font color="purple">**函数签名。**</font>定义了函数的名称和它期望的任何输入。 
### 2. <font color="purple">**函数体。**</font>包含每次使用函数时都会运行的代码。

```python
def multiply(x, y, z=0.1):   # Function signature
    #Function body
    product = x * y * z
    return product

```

```python
num = multiply(2,3,4)
num
```

```python
num = multiply(y=9, x=99)
print(num) 
```

### Python中的<font color="purple">**所有函数都会**</font>返回一个值，即使这个值是None。然而，并<font color="purple">**不是所有的函数**</font>都需要return语句

>### 2. `convert_far_to_cel()`函数将华氏度转换为摄氏度，使用以下公式: 
>### $$ C = (f - 32) * 5/9 $$
>### 该脚本应该首先提示用户输入以华氏度为单位的温度，然后显示转换为摄氏温度的温度。然后提示用户输入以摄氏度为单位的温度，并显示转换为华氏度度的温度。所有转换后的温度都应该四舍五入到小数点后两位。

```python
def convert_cel_to_far(c):
    """ 将摄氏度转换为华氏度"""
    f = c*9/5+32
    return f

def convert_far_to_cel(f):
    """将华氏度转换为摄氏度"""
    c = (f-32)*5/9
    return c

f = input("请输入一个华氏温度：")
f = float(f)
c = convert_far_to_cel(f)
print(f"华氏温度{f}转换为摄氏温度的结果是{c:.2f}")

c = input("请输入一个摄氏温度：")
c = float(c)
f = convert_cel_to_far(c)
print(f"摄氏温度{c}转换为华氏温度的结果是{f:.2f}")
```

## <font color = #0288d1>循环</font>
> ### 电脑最大的好处之一就是可以一遍又一遍地重复同样的事情。而循环则是一段代码，它会不断重复，要么达到指定的次数，要么直到满足某个条件。Python中有两种循环：<font color="purple">**while循环和for循环。**</font>先来看看while循环的工作原理。

## <font color = #0288d1>while循环</font>
+ ### while循环在某个条件为真时重复一段代码直到条件改变。每个while循环由两部分组成:
+ ### (1)while语句以<font color="purple">**while**</font>关键字开头，后面跟着一个测试条件，以冒号(:)结尾。 
+ ### (2)循环体包含在循环的每一步都会重复的代码。每行缩进4个空格。
+ ### 当执行while循环时，Python会计算测试条件，判断它是<font color="purple">**true还是false**</font>。如果测试条件为真，则执行循环体中的代码。否则，则跳过循环体中的代码，执行程序的其余部分。


```python
n = 1
while n<5:
    print(n)
    n=n+1
```

> ### 如果测试条件为真，循环体被执行，那么一旦Python到达循环体末尾，它就会返回到while语句，<font color="purple">**并重新计算测试条件**</font>。如果测试条件仍然为true，则再次执行循环体。如果为false，则<font color="purple">**跳过循环体**</font>。这个过程不断重复，直到测试条件失败。

```python
num = float(input("请输入一个正数: "))
while num <= 0:
    print("请输入正确的数")
    num = float(input("Enter a positive number: "))
```

## <font color = #0288d1>for循环</font>
>### for循环对元素集合中的每个元素执行一段代码。代码执行的<font color="purple">**次数是由集合中的数量决定**</font>。和对应的while循环一样，for循环也有两个主要部分:
>### for语句以<font color="purple">**for**</font>关键字开头，后面跟着一个成员表达式，以冒号(:)结束。 
>### 循环体包含循环每一步要执行的代码，并缩进4个空格。


```python
for letter in "Python":
    print(letter)
```

+ ### 在循环的每一步，变量letter都会被赋值给字符串“Python”中的下一个字母，然后打印letter的值。对于字符串“Python”中的每个字符，循环都会运行一次，因此循环体会执行6次。 
+ ### 为了解为什么for循环<font color="purple">**更适合处理元素的集合**</font>，我们将前面例子中的for循环重写为while循环。

```python
word = "Python"
index = 0
while index < len(word):
    print(word[index])
    index = index + 1
```

+ ### for循环经常和<font color="purple">**range函数**</font>一起使用。使用range(n)，其中n是任意正数，来精确地执行循环n次。例如，range(3)返回从0开始到3为止的整数的范围，0，1，2。<font color="darkred">**特别注意，不包括3。** </font>也就是范围是[0,n)。
+ ### 也可以给range一个起点。例如，range(1, 5)是数字1、2、3、4。第一个参数是起始数字，第二个参数是端点，它不包含在范围内。

```python
for i in range(5):
    print(i)
```

```python
amount = float(input("输入金额："))
for num in range(2, 6):
    print(f"{num}：{amount / num:,.2f} ")
```

## <font color = #0288d1>嵌套循环</font>

```python
for n in range(1, 3):
    for j in range(3, 5):
        print(f"n = {n} and j = {j}")
```

### 当Python进入第一个for循环的循环体时，变量n被赋值1。
### 然后执行第二个for循环的循环体，将j赋值为3。首先打印的是n = 1和j = 3。
### 执行`print()`函数后，Python返回到内部的for循环，将4的值赋给j，然后打印n = 1和j = 4。此时，内部的for循环执行完毕，因此控制返回到外部的for循环。变量n被赋值为2，内部的for循环再次执行。也就是说，j被赋值为3,n = 2, j = 3被打印到控制台。

### 正整数n的因子m是任何小于或等于n且n能被m除尽的数。例如，3是12的因子，因为12能够被3整除。5不是12的因子，因为12除以5的余数是2。用户输入一个正整数，然后打印出该数的因子。

```python
num = int(input("请输入一个正整数："))
for m in range(1, num+1):
    if num%m==0:
        print(f"{m}是{num}的因子")
```

### 首先设置一个初始存款，即本金金额。每年，这个金额都有一个固定的百分比增加，称为年收益率。例如，本金金额为100元，年收益率为5%，第一年增加5元，总金额为105。第二年，在第一年的基础上增加新金额的5% ，增加金额为5.25，总金额110.25。编写一个函数invest，它有三个参数:本金金额、年收益率和需要计算的年数。然后，该函数在指定年份的每年年底打印出投资金额，四舍五入到小数点后两位。

```python
def invest(amount, rate, years):
    for y in range(years):
        amount = amount*rate + amount
        print(f"第{y+1}年的总投资额为：{amount:.2f}")

amount = float(input("请输入本金："))
rate = float(input("请输入利率："))
years = int(input("请输入年限："))

invest(amount, rate, years)
```

>### 列表有一个`sort()`方法，可以对所有项目进行升序排序。默认情况下，列表是按字母或数字顺序排序的，这取决于列表中元素的类型。注意，`sort()`会<font color=darkred>**直接对列表进行排序**</font>，所以不需要将它的结果赋值给任何东西。
>### `sort()`有一个名为key的选项参数，<font color="purple">**key参数接受一个函数**</font>，可用于调整列表根据该函数的返回值进行排序。例如，要按每个字符串的长度对字符串的列表进行排序，可以将`len()`函数传递给key: 

```python
colors = ["red", "yellow", "green", "blue"]
colors.sort(key=len)
colors
```

## <font color = #0288d1>字典</font>
>### Python中最有用的数据结构之一是字典。什么是字典？Python字典像列表和元组一样，存储了一个对象的集合。然而，字典不是将对象存储在序列中，而是以数据对的形式保存信息，称为<font color="purple">**键值对**</font>。也就是说，字典中的每个对象都有两个部分：<font color="purple">**一个键和一个值。**</font>Python字典是一种将一组键与一组值联系起来的数据结构。每个键都被分配了一个值，它定义了两个集合之间的关系。

```python
# 下面的代码创建了一个字典常量，其中包含省份的名字和它们的省会城市
provincial_capital = {"陕西":"西安", "广东":"广州", "浙江":"杭州"}
```

>### 请注意，每个键与值之间用冒号(:)分隔，每个键-值对之间用逗号(,)分隔，整个字典被括在<font color="purple">**大括号({ })中。**</font>特别注意：在使用中文的时候，引号内的内容可以是中文，<font color=darkred>**但是引号本身、逗号、大括号等均应使用西文符号。**</font>
>### 最后可以有逗号，也可以不写
### 可以使用内置的dict()从一系列元组中创建字典

```python
provincial_capital["陕西"]
```

```python
key_value_pairs = (
    ("陕西", "西安"),
    ("广东", "广州"),
    ("浙江", "杭州"),
)
    
provincial_capital = dict(key_value_pairs)
```

### 要访问字典中的值，在字典末尾将对应的键用方括号([])括起来

```python
provincial_capital["陕西"]
```

### 和列表一样，字典也是<font color="purple">**可变的数据结构**。</font>这意味着你可以从字典中添加和删除元素

```python
provincial_capital["四川"]="changxing"
provincial_capital
```

### 添加时直接赋值给新的键-值对即可。字典中的每个键只能被赋予一个值。如果一个键被赋予了一个新值，Python将覆盖掉旧的值。
### 要从字典中删除项，使用<font color="purple">**del**</font>关键字，并将键指定为要删除的值:

```python
del provincial_capital["广东"]
provincial_capital
```

### 检查字典键是否存在

```python
if "陕西" in provincial_capital:
    print(provincial_capital["陕西"])
```

## <font color = #0288d1>遍历字典</font>
### 像列表和元组一样，<font color="purple">**字典也是可迭代的**</font>。然而，遍历字典与遍历列表或元组略有不同。当使用for循环遍历字典时，会遍历字典中的键:

```python
for key in provincial_capital:
    print(key)
```

```python
for province in provincial_capital:
    print(f"{province }省的省会是{provincial_capital[province]}")
```

### 使用`items()`字典方法有一种稍微简洁的方法来做到这一点，`items()`返回一个类似列表的对象，其中包含键值对的元组

```python
provincial_capital.items()
```

```python
for province, city in provincial_capital.items():
    print(f"{province}省的省会是{city}")
```
